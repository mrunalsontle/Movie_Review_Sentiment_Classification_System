```python
from keras.models import load_model
import joblib
from tensorflow.keras.preprocessing.sequence import pad_sequences

model = load_model("model.h5")
tokenizer = joblib.load("tokenizer.pkl")

def predictive_system(review):
  sequences = tokenizer.texts_to_sequences([review])
  padded_sequence = pad_sequences(sequences, maxlen=200)
  prediction = model.predict(padded_sequence)
  sentiment = "positive" if prediction[0][0] > 0.5 else "negative"
  return sentiment
```

```python
import gradio as gr
import tensorflow as tf
import pickle
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Load model and tokenizer
model = tf.keras.models.load_model("model.h5")

with open("tokenizer.pkl", "rb") as handle:
    tokenizer = pickle.load(handle)

# Preprocessing function
def preprocess_text(text, max_len=100):
    sequence = tokenizer.texts_to_sequences([text])
    padded_sequence = pad_sequences(sequence, maxlen=max_len)
    return padded_sequence

# Function to save output to a file
def save_output(movie_name, sentiment, confidence):
    with open("sentiment_results.txt", "a") as file:
        file.write(f"Movie: {movie_name} | Sentiment: {sentiment} | Confidence: {confidence:.2f}\n")

# Prediction function
def predictive_system(movie_name, review):
    if not movie_name.strip() or not review.strip():
        return "<span style='color:#FF4444; font-weight:bold;'>⚠️ Please enter both movie name and review!</span>"
    
    processed_review = preprocess_text(review)
    prediction = model.predict(processed_review)[0][0]
    
    if prediction > 0.5:
        sentiment = "<span class='sentiment-positive'>😃 Positive</span>"
    else:
        sentiment = "<span class='sentiment-negative'>😞 Negative</span>"
    
    save_output(movie_name, "Positive" if prediction > 0.5 else "Negative", prediction)
    
    return f"<strong class='label-text-black'>Movie:</strong> {movie_name}<br><strong class='label-text-black'>Sentiment:</strong> {sentiment}<br><strong class='label-text-black'>Confidence:</strong> {prediction:.2f}"

# Custom CSS
custom_css = """
.gradio-container {
    background-image: url('https://images.pexels.com/photos/7991178/pexels-photo-7991178.jpeg?auto=compress&cs=tinysrgb&w=1260&h=750&dpr=2');
    background-size: cover;
    background-position: center;
    background-repeat: no-repeat;
    background-attachment: fixed;
    color: #bbb;
    font-family: 'Poppins', sans-serif;
}

h1, h2, p {
    color: #fff !important;
    text-align: center;
    font-weight: bold;
    text-shadow: 2px 2px 4px rgba(0, 0, 0, 0.7);
}

.gr-textbox textarea {
    background: rgba(255, 255, 255, 0.1) !important;
    color: #fff !important;
    border: 1px solid #666 !important;
    border-radius: 8px !important;
    padding: 12px !important;
    font-size: 16px !important;
}

button {
    background: linear-gradient(45deg, #4A00E0, #8E2DE2) !important;
    color: white !important;
    border: none !important;
    border-radius: 8px !important;
    padding: 12px 20px !important;
    font-size: 16px !important;
    font-weight: bold !important;
    transition: all 0.3s ease-in-out !important;
    display: block;
    margin: 0 auto;
}

button:hover {
    background: linear-gradient(45deg, #8E2DE2, #4A00E0) !important;
    transform: scale(1.05);
    box-shadow: 0px 4px 10px rgba(142, 45, 226, 0.5);
}

.gr-box {
    background: rgba(0, 0, 0, 0.4) !important; /* Set to semi-transparent */
    border-radius: 12px !important;
    padding: 15px !important;
    color: #fff !important;
    font-size: 18px !important;
    text-align: center;
}

.sentiment-positive {
    color: #00FF00 !important;
    font-weight: bold !important;
    font-size: 20px !important;
}

.sentiment-negative {
    color: #FF4444 !important;
    font-weight: bold !important;
    font-size: 20px !important;
}

.label-text-black {
    color: #000000 !important;
    font-weight: bold !important;
    font-size: 18px !important;
    text-shadow: none !important;
}

"""

# Gradio UI
title = "🎬 MOVIE SENTIMENT ANALYSIS APPLICATION"
description = "Enter a movie name and its review to find out whether it is Positive or Negative!"

app = gr.Interface(
    fn=predictive_system,
    inputs=[
        gr.Textbox(label="Movie Name", placeholder="Enter movie name..."),
        gr.Textbox(label="Movie Review", placeholder="Type your review here...", lines=4)
    ],
    outputs=gr.HTML(label="Sentiment Prediction"),
    title=title,
    description=description,
    theme="compact",
    css=custom_css
)

app.launch(share=True)
```
